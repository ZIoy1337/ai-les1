### task1

```python
import random

array_1d = [random.randint(1, 100) for _ in range(10)]

print("Згенерований масив:", array_1d)

first_element = array_1d[0]

# Отримуємо центральний елемент (для масиву з 10 елементів середина — це індекс 5 або 4)
# Використовуємо цілочисельне ділення для знаходження середини: len(array_1d) // 2
middle_index = len(array_1d) // 2
middle_element = array_1d[middle_index]

# Отримуємо останній елемент (індекс -1 або len(array_1d) - 1)
last_element = array_1d[-1]

# Виводимо результати
print(f"Перший елемент: {first_element}")
print(f"Центральний елемент (індекс {middle_index}): {middle_element}")
print(f"Останній елемент: {last_element}")
```

### task3

```python
matrix_4x4 = [[i * 4 + j + 1 for j in range(4)] for i in range(4)]
print("Початкова матриця 4х4:")
for row in matrix_4x4:
    print(row)

submatrix_2x2 = [row[2:4] for row in matrix_4x4[:2]]
print("\nПідматриця 2х2 (правий верхній кут):")
for row in submatrix_2x2:
    print(row)
```

### task4

```python
import random


matrix_5x5 = [[random.randint(1, 100) for i in range(5)] for i in range(5)]
print("Матриця 5х5:")
for row in matrix_5x5:
    print(row)

center_element = matrix_5x5[2][2]

third_column = [row[2] for row in matrix_5x5]

print(f"\nЦентральний елемент: {center_element}")
print(f"Третій стовпчик: {third_column}")
```

### task6

```python
numbers = list(range(1, 21))
print("Масив чисел:", numbers)

div_by_3_count = sum(1 for x in numbers if x % 3 == 0)

print(f"Кількість елементів, які діляться на 3: {div_by_3_count}")
```

### task7

```python
import random

# Створюємо список із 12 випадкових чисел від 0 до 100
random_array = [random.randint(0, 100) for _ in range(12)]
print("Початковий масив:", random_array)

# Фільтруємо елементи, які більші за 20 і менші за 80
filtered_array = [x for x in random_array if 20 < x < 80]
print("Елементи в діапазоні (20, 80):", filtered_array)
```

### task8

```python
import random

# Створюємо 2D масив 3х4 із випадкових чисел
matrix_3x4 = [[random.randint(1, 20) for _ in range(4)] for _ in range(3)]
print("Початковий 2D масив 3х4:")
for row in matrix_3x4:
    print(row)

# Замінюємо парні числа на 0, а непарні залишаємо
modified_matrix = [[(0 if x % 2 == 0 else x) for x in row] for row in matrix_3x4]

print("\nМасив після заміни парних чисел на 0:")
for row in modified_matrix:
    print(row)
```
