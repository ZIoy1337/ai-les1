### Task 1

```python
temperatures = [2, 5, 4, 6, 7, 3, 1]

mean_temp = sum(temperatures) / len(temperatures)
max_temp = max(temperatures)
min_temp = min(temperatures)
diff_temp = max_temp - min_temp

print(f"Середнє значення: {mean_temp}")
print(f"Максимальне значення: {max_temp}")
print(f"Мінімальне значення: {min_temp}")
print(f"Різниця між максимальним і мінімальним: {diff_temp}")
```

### Task 2

```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]

total_expenses = sum(expenses)
mean_expenses = total_expenses / len(expenses)
max_expense = max(expenses)
min_expense = min(expenses)

days_above_1000 = sum(1 for x in expenses if x > 1000)

print(f"Загальна сума витрат: {total_expenses}")
print(f"Середнє значення: {mean_expenses}")
print(f"Максимальна витрата: {max_expense}")
print(f"Мінімальна витрата: {min_expense}")
print(f"Кількість днів > 1000: {days_above_1000}")
```

### Task 3

```python
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]

vote_counts = {}
for fruit in votes:
    vote_counts[fruit] = vote_counts.get(fruit, 0) + 1

leader = max(vote_counts, key=vote_counts.get)
leader_votes = vote_counts[leader]

sorted_votes = sorted(vote_counts.items(), key=lambda x: x[1], reverse=True)
top_3 = sorted_votes[:3]

print(f"Лідер голосування: {leader} ({leader_votes} голосів)")
print("Топ-3:", top_3)
```

### Task 4

```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Math": math,
    "Physics": physics,
    "English": english
}

results = {}
for subject, scores in subjects.items():
    avg_score = sum(scores) / len(scores)
    min_score = min(scores)
    max_score = max(scores)
    results[subject] = {"avg": avg_score, "min": min_score, "max": max_score}
    print(f"{subject} -> Середній: {avg_score}, Мін: {min_score}, Макс: {max_score}")

best_subject = max(results, key=lambda k: results[k]["avg"])
print(f"Предмет з найвищим середнім результатом: {best_subject}")
```
