### task1

```python
import pandas as pd

# 1. Створюємо словник із даними про книги (не менше 6 рядків)
data = {
    "Назва": [
        "Clean Code",
        "The Pragmatic Programmer",
        "Designing Data-Intensive Applications",
        "Python Crash Course",
        "Refactoring",
        "You Don't Know JS"
    ],
    "Автор": [
        "Robert C. Martin",
        "Andrew Hunt, David Thomas",
        "Martin Kleppmann",
        "Eric Matthes",
        "Martin Fowler",
        "Kyle Simpson"
    ],
    "Рік видання": [2008, 2019, 2017, 2019, 2018, 2015],
    "Ціна": [850, 1150, 1300, 750, 1200, 600]
}

df = pd.DataFrame(data)

print("--- Увесь DataFrame ---")
print(df)
print("\n" + "="*50 + "\n")

average_price = df["Ціна"].mean()
print(f"Середня ціна книг: {average_price:.2f} грн")
print("\n" + "="*50 + "\n")

books_after_2015 = df[df["Рік видання"] > 2015]
print("--- Книги, видані після 2015 року ---")
print(books_after_2015)
print("\n" + "="*50 + "\n")

df_sorted = df.sort_values(by="Ціна", ascending=True)
print("--- Відсортовано за ціною (зростання) ---")
print(df_sorted)
```

### task2

```python
import pandas as pd


df_orders = pd.read_csv('orders.csv')

print("--- Перші 10 рядків замовлень ---")
print(df_orders.head(10))
print("\n" + "="*50 + "\n")

orders_per_client = df_orders['Клієнт'].value_counts()
print("--- Кількість замовлень кожного клієнта ---")
print(orders_per_client)
print("\n" + "="*50 + "\n")

max_amount = df_orders['Сума'].max()
min_amount = df_orders['Сума'].min()
print(f"Максимальна сума замовлення: {max_amount}")
print(f"Мінімальна сума замовлення: {min_amount}")
print("\n" + "="*50 + "\n")

total_sum = df_orders['Сума'].sum()
print(f"Загальна сума всіх замовлень: {total_sum}")
```

### task3

```python
import pandas as pd

clients_data = {
    "ID": [1, 2, 3, 4, 5],
    "Ім'я": ["Олександр", "Марія", "Іван", "Софія", "Андрій"],
    "Місто": ["Київ", "Львів", "Одеса", "Харків", "Дніпро"]
}
df_clients = pd.DataFrame(clients_data)

orders_data = {
    "ID клієнта": [1, 2, 1, 3, 5],
    "Замовлення": ["Ноутбук", "Мишка", "Клавіатура", "Монітор", "Навушники"],
    "Вартість": [25000, 800, 1500, 7000, 2000]
}
df_orders_db = pd.DataFrame(orders_data)

print("--- Таблиця клієнтів ---")
print(df_clients)
print("\n" + "="*50 + "\n")

print("--- Таблиця замовлень ---")
print(df_orders_db)
print("\n" + "="*50 + "\n")

merged_df = pd.merge(df_orders_db, df_clients, left_on="ID клієнта", right_on="ID")

result_df = merged_df.groupby(["ID клієнта", "Ім'я", "Місто"])["Вартість"].sum().reset_index()

result_df = result_df.rename(columns={"Вартість": "Загальна вартість замовлень"})

print("--- Результат: Загальна вартість замовлень за кожним клієнтом ---")
print(result_df)
```
